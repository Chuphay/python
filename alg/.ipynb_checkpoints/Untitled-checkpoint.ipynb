```python
def combinations(n):
    if n == 1:
        return ["1"]
    if n == 2:
        return ["2","11"]
    tmp = []
    two_back = combinations(n-2)
    for e in two_back:
        tmp.append(e + "2")
        tmp.append("2" + e)
    one_back = combinations(n-1)
    for e in one_back:
        tmp.append(e + "1")
        tmp.append("1"+ e)
        
    return list(set(tmp))
```

```python
combinations(3)
```
